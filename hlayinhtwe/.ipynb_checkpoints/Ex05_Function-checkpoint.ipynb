Ex 1: Write a function called rectangle that takes two integers m and n as arguments and prints
out an m× n box consisting of asterisks. Shown below is the output of rectangle(2,4)
****
****

```python
def rectangle(m,n):
    for k in range(m):
        print(n*'*')

rectangle(2,4)
```

Ex 2: Write a function that takes an integer n and returns a random integer with exactly n digits. For
instance, if n is 3, then 125 and 593 would be valid return values, but 093 would not because
that is really 93, which is a two-digit number.

```python
#if n=3 => 100 and 999 //smallest to largest
import random
def random_dig(n):
    x = random.randint(10**(n-1), 10**n-1)
    
    return x
    
print(random_dig(3))
```

Ex 3: Write a function called number_of_factors that takes an integer and returns how many
factors the number has.

```python
#1,2,3,4,6,12=> 12
def number_of_factors(N):
    # result=n % 
    # return result
    factor_list = []
    for d in range(1,N+1):
        if(N%d ==0):
            factor_list.append(d)
    return len(factor_list)

number_of_factors(12)
    
    
```

Ex4: Write a function called first_diff that is given two strings and returns the first location in
which the strings differ. If the strings are identical, it should return -1.Ignore the case in your function. It means h and H are the same.

```python

def first_diff(s1,s2):
    s1 = s1.upper()
    s2 = s2.upper()
    if(s1==s2):
        print('two strings are identical')
        diff = -1
    else:
        diff = min(len(s1), len(s2))
        for idx in range(min(len(s1), len(s2))):            
            if not (s1[idx]==s2[idx]):
                diff = idx
                break
    return diff         

print('case 0:')
first_diff('Hello', 'hello')
print('case 1: same length - diff at index', first_diff('Hello', 'hellp'))
print('case 2: diff length - diff at index', first_diff('Hel', 'hellp'))
print('case 3: diff length - diff at index', first_diff('Hello', 'hel'))
print('case 4: same length, all diff - diff at index', first_diff('Hello', 'arr'))
```

```python
def first_diff(a,b):
    a= a.lower()
    b= b.lower()

    for i in range (min(len(a) , len(b))):
        if a[i] != b[i]:
            return i

    if len(a) != len(b):
        return min(len(a) , len(b))
        
    return -1
            
print(first_diff("hello", "Hello"))
print(first_diff("Hello", "hello") )    # ➜ -1
print(first_diff("Apple", "apply"))    # ➜ 4
print(first_diff("MMDT", "mmdtPython") )      # ➜ 2
print(first_diff("Dog", "Dogs"))         # ➜ 3
```

Ex5: Write a function called root that is given a number x and an integer n and returns x/n. In
the function definition, set the default value of n to 2.

```python
#The original code/question has a logical error. 

def root(x=1, n=2):
    try:    
        assert x > 0, "base x must be greater than zero."
        result = x**(1/n)
        return result
    except ZeroDivisionError as e:
        print(f'Error: {e}')
    except TypeError as e:
        print(f'Error: {e}')
    except ValueError as e:
        print(f'Error: {e}')
    except OverflowError as e:
        print(f'Error: {e}')
    except AssertionError as e:
        print(f'Error: {e}')
    except MemoryError as e:
        print(f'Error: {e}')
    else:
        print('Other Errors')
    finally: 
        print('Code is Complete here!')


root(-8, 3)
root(8, 0)
root(10**1000, 3)
root('hello', 3)
```

Ex6: Write a function that solves the system of equations ax + by = e and cx +dy = f. Your function should takes 6 parameters, (a, b, c, d, e,f) and return the values of x and y.

```python
 # ax + by = e and cx +dy = f
def solve_system(a, b, c, d, e, f):
    determinant = a * d - b * c

    if determinant == 0:
        return None   # No unique solution

    x = (e * d - b * f) / determinant
    y = (a * f - e * c) / determinant

    return x, y

solve_system(2,3,4,5,6,7)
```

Ex7:  Write a function called change_case that given a string, returns a string with each upper  case letter replaced by a lower case letter and vice-versa

```python
def change_case(text):
    result=""
    
    for char in text:
        if char.isupper():
            result = result + char.lower()
        elif char.islower():
            result = result + char.upper()
        else:
            result = result+ char
    return result


print(change_case("Hello World") )
print(change_case("PyThOn123") )
```

Ex8:  Write a function called primes that is given a number n and returns a list of the first n  primes. Let the default value of n be 100.

```python
def primes(n=100):
    prime_list = [] 
    num = 2 
    while len(prime_list) < n:
        is_prime = True
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            prime_list.append(num)

        num += 1

    return prime_list
print(primes(10))
```

Ex10:  Write a function that is given a 9 × 9 potentially solved Sudoku and returns True if it is solved correctly and False if there is a mistake. The Sudoku is correctly solved if there are no repeated numbers in any row or any column or in any of the nine “blocks.”


```python
def is_sudoku_solved(sudoku):
    correct = set(range(1, 10))

    if any(set(row) != correct for row in sudoku):
        return False

    if any(set(sudoku[r][c] for r in range(9)) != correct for c in range(9)):
        return False

    for block_row in range(0, 9, 3):
        for block_col in range(0, 9, 3):
            block = {sudoku[r][c] for r in range(block_row, block_row+3)
                               for c in range(block_col, block_col+3)}
            if block != correct:
                return False

    return True
sudoku = [
    [5,3,4,6,7,8,9,1,2],
    [6,7,2,1,9,5,3,4,8],
    [1,9,8,3,4,2,5,6,7],
    [8,5,9,7,6,1,4,2,3],
    [4,2,6,8,5,3,7,9,1],
    [7,1,3,9,2,4,8,5,6],
    [9,6,1,5,3,7,2,8,4],
    [2,8,7,4,1,9,6,3,5],
    [3,4,5,2,8,6,1,7,9]
]

print(is_sudoku_solved(sudoku)) 
```
